### **Ejercicio 1. Control de calidad**
Una empresa de componentes electrónicos revisa 200 piezas: 18 son defectuosas y 182 conformes.

Actividades:

a) Halla P(defectuosa).

b) Halla P(conforme).

c) Interpreta ambos porcentajes.

```python
total = 200
defectuosas = 18
conformes = 182
p_defectuosa = defectuosas / total
p_conforme = conformes / total
print(f"a) P(defectuosa) = {p_defectuosa:.2f} = {p_defectuosa:.0%}")
print(f"b) P(conforme) = {p_conforme:.2f} = {p_conforme:.0%}")
```

### Mi procedimiento

Primero tengo los datos: hay 200 piezas, de estas 18 son defectuosas y 182 están conformes.

Para encontrar la probabilidad, divido la cantidad que me interesa entre el total. Esta es la fórmula:

P(evento) = casos favorables / casos totales.

**a) Probabilidad de una pieza defectuosa**

P(defectuosa) = 18 / 200 = **0.09**

Para pasarlo a porcentaje, multiplico por 100:

0.09 × 100 = **9 %**

**b) Probabilidad de una pieza conforme**

P(conforme) = 182 / 200 = **0.91**

En porcentaje:

0.91 × 100 = **91 %**

**c) Interpretación**

Entiendo que, al escoger al azar una de estas piezas, tengo un 9 % de probabilidad de sacar una defectuosa y un 91 % de sacar una conforme.

P(puntual) = 425 / 500 = **0.85**

Primero resto los pedidos puntuales del total:

En porcentaje:

### **Ejercicio 5. Tasa de conversión digital**
Cada visita a una página tiene una probabilidad de compra de 0.20. Supón 3 visitas independientes y
la misma probabilidad para cada una.

Actividades:

a) Calcula P(exactamente 1 compra).

b) P(al menos 1 compra).

c) Explica el supuesto de independencia.

```python
from math import comb
n, p = 3, 0.20
p_una = comb(n, 1) * p * (1 - p)**2
p_ninguna = (1 - p)**n
p_al_menos_una = 1 - p_ninguna
print(f"a) P(exactamente 1 compra) = {p_una:.3f} = {p_una:.1%}")
print(f"b) P(al menos 1 compra) = {p_al_menos_una:.3f} = {p_al_menos_una:.1%}")
```

### Mi procedimiento

**Datos:**

Número de visitas: n = 3

Probabilidad de compra: p = 0.20

Probabilidad de no comprar: 1 − 0.20 = 0.80

X = cantidad de visitas en las que hay una compra

**a) Probabilidad de exactamente una compra**

Uso la fórmula binomial: P(X = k) = C(n,k) × p^k × (1 − p)^(n − k).

C(n,k) cuenta las formas de repartir las compras entre las visitas. Como quiero una compra entre tres visitas:

C(3,1) = 3! / (1! × 2!) = **3**

La única compra puede ser en la primera, segunda o tercera visita. Reemplazo los datos:

P(X = 1) = 3 × 0.20 × 0.80²

P(X = 1) = 3 × 0.20 × 0.64 = **0.384**

Lo paso a porcentaje:

0.384 × 100 = **38.4 %**

**b) Probabilidad de al menos una compra**

“Al menos una” significa una, dos o tres compras. Primero calculo la probabilidad de que no haya ninguna:

P(X = 0) = 0.80³ = **0.512**

Después resto ese resultado de 1:

P(X ≥ 1) = 1 − P(X = 0) = 1 − 0.512 = **0.488**

En porcentaje: 0.488 × 100 = **48.8 %**

**c) Supuesto de independencia**

Entiendo que lo que pase en una visita no cambia la probabilidad de compra de las otras. Además, la probabilidad de compra sigue siendo 0.20 en cada visita.

### **Ejercicio 6. Entregas más puntuales**
Históricamente, 80 % de los pedidos de una empresa llegan a tiempo. Luego de una mejora, 45 de
50 pedidos llegan puntualmente. Se plantea H0: p = 0.80 frente a H1: p > 0.80.

Actividades:

a) Identifica la variable y los parámetros.

b) Usa una prueba binomial exacta y determina el p-valor.

c) Decide con α = 0.05.

```python
from math import comb
n, x, p0, alfa = 50, 45, 0.80, 0.05
print("X = cantidad de pedidos puntuales entre 50")
print(f"n = {n}, x = {x}, p0 = {p0}, proporción observada = {x/n:.2f}")
# Sumo desde lo observado hasta el mayor resultado posible.
p_valor = 0
for k in range(x, n + 1):
    probabilidad = comb(n, k) * p0**k * (1 - p0)**(n - k)
    p_valor += probabilidad
    print(f"P(X = {k}) = {probabilidad:.8f}")
print(f"p-valor = {p_valor:.8f}")
print("Rechazo H0" if p_valor <= alfa else "No rechazo H0")
```

### Mi procedimiento

**Datos:**

Total: n = 50 pedidos

Pedidos puntuales observados: x = 45

Proporción histórica: p₀ = 0.80

Nivel de significancia: α = 0.05

H₀: p = 0.80. H₁: p > 0.80.

**a) Variable y parámetros**

Llamo X a la cantidad de pedidos puntuales entre los 50. El parámetro p es la proporción real de pedidos que llegan a tiempo después de la mejora.

Calculo la proporción que observé:

p̂ = 45 / 50 = 0.90 = **90 %**

Para la prueba supongo que H₀ es cierta. Uso una binomial con n = 50 y p₀ = 0.80, suponiendo pedidos independientes y la misma probabilidad de puntualidad.

**b) Cálculo del p-valor**

Como quiero saber si la puntualidad aumentó, calculo la probabilidad de obtener 45 pedidos puntuales o más:

p-valor = P(X ≥ 45) = P(X = 45) + … + P(X = 50).

Uso la fórmula: P(X = k) = C(50,k) × 0.80^k × 0.20^(50 − k).

Por ejemplo, para 45 pedidos puntuales:

P(X = 45) = C(50,45) × 0.80^45 × 0.20^5 ≈ 0.02953120.

Calculo también los casos de 46 a 50 y sumo:

p-valor ≈ 0.02953120 + 0.01283965 + 0.00437095 + 0.00109274 + 0.00017841 + 0.00001427.

p-valor ≈ **0.04802722**

**c) Decisión con α = 0.05**

Uso la regla: si p-valor ≤ α, rechazo H₀.

Comparo: **0.04802722 < 0.05**. Por eso rechazo H₀.

Encuentro evidencia de que la proporción de pedidos puntuales aumentó por encima del 80 %. El resultado está cerca del límite de 0.05; no significa que la mejora sea segura.

### **Ejercicio 7. Mejora en calidad**
Un fabricante afirma que su tasa de defectos es 10 %. En una muestra de 40 piezas se observan 1
defectuosa. Contrasta H₀: p = 0.10 frente a H₁: p < 0.10.

Actividades:

a) Calcula la cola inferior binomial exacta.

b) Decide con α = 0.05.

c) Redacta una conclusión.

```python
from math import comb
n, x, p0, alfa = 40, 1, 0.10, 0.05
print(f"Proporción observada = {x/n:.3f} = {x/n:.1%}")
# Para la cola inferior incluyo cero y una pieza defectuosa.
p_valor = 0
for k in range(x + 1):
    probabilidad = comb(n, k) * p0**k * (1 - p0)**(n - k)
    p_valor += probabilidad
    print(f"P(X = {k}) = {probabilidad:.8f}")
print(f"p-valor = {p_valor:.8f}")
print("Rechazo H0" if p_valor <= alfa else "No rechazo H0")
```

### Mi procedimiento

**Datos:**

Total: n = 40 piezas

Piezas defectuosas observadas: x = 1

Tasa de defectos indicada: p₀ = 0.10

Nivel de significancia: α = 0.05

H₀: p = 0.10. H₁: p < 0.10.

**a) Cálculo de la cola inferior binomial**

Llamo X al número de piezas defectuosas. Primero calculo la proporción observada:

p̂ = 1 / 40 = 0.025 = **2.5 %**

Como quiero saber si la tasa bajó, sumo la probabilidad de encontrar cero o una pieza defectuosa, suponiendo cierta H₀:

p-valor = P(X ≤ 1) = P(X = 0) + P(X = 1).

Uso la fórmula: P(X = k) = C(40,k) × 0.10^k × 0.90^(40 − k).

Para ninguna pieza defectuosa:

P(X = 0) = C(40,0) × 0.10⁰ × 0.90⁴⁰ = 0.90⁴⁰ ≈ **0.01478088**.

Para una pieza defectuosa:

P(X = 1) = C(40,1) × 0.10 × 0.90³⁹ = 40 × 0.10 × 0.90³⁹ ≈ **0.06569281**.

Sumo los dos resultados:

p-valor ≈ 0.01478088 + 0.06569281 ≈ **0.08047370**.

El resultado final está redondeado después de sumar con todos los decimales. Este modelo supone piezas independientes y la misma probabilidad de defecto.

**b) Decisión con α = 0.05**

Uso la regla: si p-valor ≤ α, rechazo H₀.

Comparo: **0.08047370 > 0.05**. Por eso no rechazo H₀.

**c) Conclusión**

Aunque observé solo un 2.5 % de piezas defectuosas, no tengo evidencia suficiente al nivel del 5 % para concluir que la tasa real bajó del 10 %. Esto tampoco demuestra que la tasa sea exactamente del 10 %.

### **Ejercicio 9. Campaña de ventas**
En una prueba de hipótesis bilateral H0: p = 0.50 frente a H1: p ≠ 0.50, se reportó p-valor = 0.081.

Actividades:

a) Decide con α = 0.05.

b) Decide con α = 0.10.

c) ¿Por qué cambian las decisiones?

```python
# Este p-valor ya es bilateral: no hay que duplicarlo.
p_valor = 0.081
for alfa in [0.05, 0.10]:
    decision = "Rechazo H0" if p_valor <= alfa else "No rechazo H0"
    print(f"Con alfa = {alfa:.2f} y p-valor = {p_valor}: {decision}")
```

### Mi procedimiento

**Datos:**

H0: p = 0.50

H1: p ≠ 0.50

p-valor = 0.081

Niveles que voy a comparar: α = 0.05 y α = 0.10

Aquí p representa la proporción que se estudia. Como el p-valor ya es de una prueba bilateral, lo uso tal como está; no lo multiplico por dos.

**a) Decisión con α = 0.05**

Uso la regla: si p-valor ≤ α, rechazo H₀.

Comparo: **0.081 > 0.05**.

No rechazo H0. Al nivel del 5 %, no tengo evidencia suficiente para decir que la proporción sea distinta de 0.50. Esto no demuestra que sea exactamente 0.50.

**b) Decisión con α = 0.10**

Ahora comparo el mismo p-valor con el otro límite:

**0.081 < 0.10**

Rechazo H0. Al nivel del 10 %, sí encuentro evidencia de que la proporción sea diferente de 0.50.

**c) Por qué cambian las decisiones**

Observo que el p-valor queda entre los dos límites:

**0.05 < 0.081 < 0.10**

Con 0.10 es más fácil rechazar H0, pero también se permite un mayor riesgo de rechazarla cuando es cierta. Los datos no cambiaron; cambió el límite de comparación. Ese límite debe elegirse antes de mirar el resultado.
