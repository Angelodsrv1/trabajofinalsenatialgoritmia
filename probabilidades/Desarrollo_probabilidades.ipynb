Ejercicio 1. Control de calidad
Una empresa de componentes electrónicos revisa 200 piezas: 18 son defectuosas y 182 conformes.

Actividades:

a) Halla P(defectuosa).

b) Halla P(conforme).

c) Interpreta ambos porcentajes.

```python
total = 200
defectuosas = 18
conformes = 182
p_defectuosa = defectuosas / total
p_conforme = conformes / total
print(f"a) P(defectuosa) = {p_defectuosa:.2f} = {p_defectuosa:.0%}")
print(f"b) P(conforme) = {p_conforme:.2f} = {p_conforme:.0%}")
```

### Mi procedimiento

Primero anoto los datos: hay 200 piezas, de las cuales 18 son defectuosas y 182 están conformes.

Para encontrar cada probabilidad, divido la cantidad que me interesa entre el total:

P(evento) = casos favorables / casos totales.

a) P(defectuosa) = 18 / 200 = 0.09.

Para pasarlo a porcentaje: 0.09 × 100 = 9 %.

b) P(conforme) = 182 / 200 = 0.91.

En porcentaje: 0.91 × 100 = 91 %.

c) Entiendo que, al escoger al azar una de estas piezas, tengo un 9 % de probabilidad de sacar una defectuosa y un 91 % de sacar una conforme. Compruebo que 9 % + 91 % = 100 %.

Ejercicio 2. Entregas a tiempo
De 500 pedidos, 425 llegaron dentro del plazo establecido. Se elige al azar uno de esos pedidos.

Actividades:

a) Calcula la probabilidad de entrega puntual.

b) La de entrega tardía.

c) Explica el resultado.

```python
total = 500
puntuales = 425
tardios = total - puntuales
print("Pedidos tardíos:", tardios)
print(f"a) P(puntual) = {puntuales / total:.2f} = {puntuales / total:.0%}")
print(f"b) P(tardía) = {tardios / total:.2f} = {tardios / total:.0%}")
```

### Mi procedimiento

Anoto que son 500 pedidos y que 425 llegaron a tiempo. Primero averiguo cuántos llegaron tarde:

Pedidos tardíos = 500 − 425 = 75.

Uso la fórmula P(evento) = cantidad de pedidos del evento / total de pedidos.

a) P(puntual) = 425 / 500 = 0.85 = 85 %.

b) P(tardía) = 75 / 500 = 0.15 = 15 %.

c) Si elijo al azar uno de esos 500 pedidos, la probabilidad de que haya llegado a tiempo es del 85 % y la de que haya llegado tarde es del 15 %. En este grupo, la mayoría llegó dentro del plazo.

excluyentes.

Actividades:

a) Calcula P(red o software).

b) P(no hardware).

c) Explica los supuestos.

```python
total = 120
red, software, hardware = 50, 40, 30
p_red_o_software = (red + software) / total
p_no_hardware = 1 - hardware / total
print(f"a) P(red o software) = {p_red_o_software:.2f} = {p_red_o_software:.0%}")
print(f"b) P(no hardware) = {p_no_hardware:.2f} = {p_no_hardware:.0%}")
```

### Mi procedimiento

Tengo 120 tickets: 50 de red, 40 de software y 30 de hardware.

a) Como un ticket no está en dos categorías a la vez, sumo los de red y software:

Tickets de red o software = 50 + 40 = 90.

P(red o software) = P(red) + P(software).

P(red o software) = 50/120 + 40/120 = 90/120 = 0.75 = 75 %.

b) Para “no hardware”, quito los 30 de hardware del total:

P(no hardware) = 1 − P(hardware).

P(no hardware) = 1 − 30/120 = 90/120 = 0.75 = 75 %.

c) Estoy suponiendo que elijo un ticket al azar y que todos tienen la misma oportunidad de salir. También tomo en cuenta que las categorías no se mezclan y que entre las tres incluyen los 120 tickets. Por eso, “no hardware” equivale aquí a “red o software”. No necesito suponer independencia entre las categorías.

Ejercicio 4. Autenticación y probabilidad condicional
En 300 intentos de acceso, 180 fueron desde móviles y 120 desde computadoras. Hubo 27 accesos
fallidos desde móviles y 12 desde computadoras.

Actividades:

a) Halla P(fallo | móvil).

b) P(fallo | computadora).

c) Compara.

### Mi procedimiento

Anoto los datos: 180 intentos desde móviles, con 27 fallos, y 120 desde computadoras, con 12 fallos.

La barra “|” la leo como “sabiendo que”. Por eso uso como total solamente los intentos del dispositivo que me preguntan.

Fórmula: P(fallo | dispositivo) = fallos del dispositivo / intentos del dispositivo.

a) P(fallo | móvil) = 27 / 180 = 0.15 = 15 %.

b) P(fallo | computadora) = 12 / 120 = 0.10 = 10 %.

c) Comparo: 15 % − 10 % = 5 puntos porcentuales.

En estos datos, la proporción de fallos fue mayor desde móviles. La diferencia fue de 5 puntos porcentuales. Esto por sí solo no me dice cuál fue la causa.

Ejercicio 5. Tasa de conversión digital
Cada visita a una página tiene una probabilidad de compra de 0.20. Supón 3 visitas independientes y
la misma probabilidad para cada una.

Actividades:

a) Calcula P(exactamente 1 compra).

b) P(al menos 1 compra).

c) Explica el supuesto de independencia.

```python
from math import comb
n, p = 3, 0.20
p_una = comb(n, 1) * p * (1 - p)**2
p_ninguna = (1 - p)**n
p_al_menos_una = 1 - p_ninguna
print(f"a) P(exactamente 1 compra) = {p_una:.3f} = {p_una:.1%}")
print(f"b) P(al menos 1 compra) = {p_al_menos_una:.3f} = {p_al_menos_una:.1%}")
```

### Mi procedimiento

Anoto n = 3 visitas y p = 0.20 de probabilidad de compra. La probabilidad de no comprar es 1 − 0.20 = 0.80.

Llamo X a la cantidad de visitas en las que hay compra. Uso el modelo binomial, que cuenta cuántos éxitos hay en un número fijo de intentos.

a) La fórmula es P(X = k) = C(n,k) × p^k × (1 − p)^(n − k).

C(n,k) cuenta las formas de colocar las compras entre las visitas.

Para una compra: C(3,1) = 3! / (1! × 2!) = 3.

P(X = 1) = 3 × 0.20 × 0.80².

P(X = 1) = 3 × 0.20 × 0.64 = 0.384 = 38.4 %.

También lo entiendo así: la única compra puede ocurrir en la primera, la segunda o la tercera visita.

b) “Al menos una” incluye una, dos o tres compras. Me resulta más fácil restar el caso de ninguna compra:

P(X ≥ 1) = 1 − P(X = 0).

P(X = 0) = 0.80³ = 0.512.

P(X ≥ 1) = 1 − 0.512 = 0.488 = 48.8 %.

c) La independencia significa que comprar o no comprar en una visita no cambia la probabilidad de compra de las otras. Además, uso la misma probabilidad de 0.20 en las tres visitas.

Ejercicio 6. Entregas más puntuales
Históricamente, 80 % de los pedidos de una empresa llegan a tiempo. Luego de una mejora, 45 de
50 pedidos llegan puntualmente. Se plantea H₀: p = 0.80 frente a H₁: p > 0.80.

Actividades:

a) Identifica la variable y los parámetros.

b) Usa una prueba binomial exacta y determina el p-valor.

c) Decide con α = 0.05.

```python
from math import comb
n, x, p0, alfa = 50, 45, 0.80, 0.05
print("X = cantidad de pedidos puntuales entre 50")
print(f"n = {n}, x = {x}, p0 = {p0}, proporción observada = {x/n:.2f}")
# Sumo desde lo observado hasta el mayor resultado posible.
p_valor = 0
for k in range(x, n + 1):
    probabilidad = comb(n, k) * p0**k * (1 - p0)**(n - k)
    p_valor += probabilidad
    print(f"P(X = {k}) = {probabilidad:.8f}")
print(f"p-valor = {p_valor:.8f}")
print("Rechazo H0" if p_valor <= alfa else "No rechazo H0")
```

### Mi procedimiento

a) Mi variable X es la cantidad de pedidos que llegan a tiempo entre los 50 revisados. Cada pedido puede llegar puntual o no.

Anoto n = 50, x = 45, p₀ = 0.80 y α = 0.05.

El parámetro p es la proporción real de pedidos puntuales después de la mejora. Mi proporción observada es p̂ = 45/50 = 0.90 = 90 %.

H₀: p = 0.80. H₁: p > 0.80.

b) Para calcular, supongo por un momento que H₀ es cierta. Entonces X sigue una binomial con n = 50 y p = 0.80. Este modelo supone pedidos independientes y la misma probabilidad de puntualidad.

Como busco un aumento, sumo la probabilidad de obtener 45 pedidos puntuales o más:

p-valor = P(X ≥ 45) = P(X = 45) + … + P(X = 50).

Uso P(X = k) = C(50,k) × 0.80^k × 0.20^(50 − k).

Por ejemplo, P(X = 45) = C(50,45) × 0.80^45 × 0.20^5 ≈ 0.02953120.

Al calcular los seis términos, obtengo:

p-valor ≈ 0.02953120 + 0.01283965 + 0.00437095 + 0.00109274 + 0.00017841 + 0.00001427.

p-valor ≈ 0.04802722.

c) Comparo: 0.04802722 < 0.05. Por eso rechazo H₀.

Con este nivel de significancia encuentro evidencia de que la proporción de entregas puntuales aumentó por encima del 80 %. La decisión queda cerca del límite de 0.05; no significa que la mejora esté demostrada con total seguridad.

Ejercicio 7. Mejora en calidad
Un fabricante afirma que su tasa de defectos es 10 %. En una muestra de 40 piezas se observan 1
defectuosa. Contrasta H₀: p = 0.10 frente a H₁: p < 0.10.

Actividades:

a) Calcula la cola inferior binomial exacta.

b) Decide con α = 0.05.

c) Redacta una conclusión.

```python
from math import comb
n, x, p0, alfa = 40, 1, 0.10, 0.05
print(f"Proporción observada = {x/n:.3f} = {x/n:.1%}")
# Para la cola inferior incluyo cero y una pieza defectuosa.
p_valor = 0
for k in range(x + 1):
    probabilidad = comb(n, k) * p0**k * (1 - p0)**(n - k)
    p_valor += probabilidad
    print(f"P(X = {k}) = {probabilidad:.8f}")
print(f"p-valor = {p_valor:.8f}")
print("Rechazo H0" if p_valor <= alfa else "No rechazo H0")
```

### Mi procedimiento

Anoto n = 40 piezas, x = 1 defectuosa, p₀ = 0.10 y α = 0.05.

Llamo X al número de piezas defectuosas. La proporción observada es p̂ = 1/40 = 0.025 = 2.5 %.

H₀: p = 0.10. H₁: p < 0.10.

a) Como quiero saber si la tasa bajó, calculo la probabilidad de encontrar una defectuosa o ninguna, suponiendo cierta H₀:

p-valor = P(X ≤ 1) = P(X = 0) + P(X = 1).

Uso P(X = k) = C(40,k) × 0.10^k × 0.90^(40 − k).

P(X = 0) = C(40,0) × 0.10⁰ × 0.90⁴⁰ = 0.90⁴⁰ ≈ 0.01478088.

P(X = 1) = C(40,1) × 0.10¹ × 0.90³⁹ = 40 × 0.10 × 0.90³⁹ ≈ 0.06569281.

p-valor ≈ 0.01478088 + 0.06569281 = 0.08047370.

Para usar este modelo, supongo piezas independientes y la misma probabilidad de defecto.

b) Comparo: 0.08047370 > 0.05. No rechazo H₀.

c) Aunque en la muestra vi solo un 2.5 % de piezas defectuosas, no tengo evidencia suficiente, con α = 0.05, para concluir que la tasa real bajó del 10 %. Tampoco puedo afirmar que la tasa sea exactamente del 10 %.

Ejercicio 9. Campaña de ventas
En una prueba de hipótesis bilateral H₀: p = 0.50 frente a H₁: p ≠ 0.50, se reportó p-valor = 0.081.

Actividades:

a) Decide con α = 0.05.

b) Decide con α = 0.10.

c) ¿Por qué cambian las decisiones?

```python
# Este p-valor ya es bilateral: no hay que duplicarlo.
p_valor = 0.081
for alfa in [0.05, 0.10]:
    decision = "Rechazo H0" if p_valor <= alfa else "No rechazo H0"
    print(f"Con alfa = {alfa:.2f} y p-valor = {p_valor}: {decision}")
```

### Mi procedimiento

Anoto H₀: p = 0.50, H₁: p ≠ 0.50 y p-valor = 0.081. Aquí p es la proporción que se está estudiando.

Como el p-valor ya corresponde a una prueba bilateral, lo uso tal como está; no lo multiplico por dos.

Mi regla es rechazar H₀ cuando p-valor ≤ α.

a) Con α = 0.05: 0.081 > 0.05.

No rechazo H₀. No tengo evidencia suficiente al nivel del 5 % para concluir que la proporción sea diferente de 0.50.

b) Con α = 0.10: 0.081 < 0.10.

Rechazo H₀. Al nivel del 10 %, sí encuentro evidencia de que la proporción es distinta de 0.50.

c) Las decisiones cambian porque 0.081 queda entre 0.05 y 0.10. El límite de 0.10 es menos exigente para rechazar H₀ y permite un mayor riesgo de rechazarla cuando es cierta. El resultado de la prueba es el mismo; lo que cambia es el límite de comparación. Ese límite debe elegirse antes de mirar el resultado.
